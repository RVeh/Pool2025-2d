## Konfidenzintervalle – ein Baukasten

Die Materialien verbinden zwei Zugänge:

| GeoGebra | Python |
|:--|:--|
| schnell entdecken und variieren | transparent, reproduzierbar und gezielt verändern |

**GeoGebra öffnet den Raum zum Entdecken. Python macht den Weg transparent und reproduzierbar.**

Die Dateien können einzeln verwendet oder zu einem Lernweg verbunden werden. Die folgende Reihenfolge ist ein Vorschlag – keine Vorgabe.

## Arbeiten in JupyterLite

1. Ein Notebook über den Link öffnen.
2. Nur die Zelle **Eingaben** verändern.
3. Anschließend alle Zellen von oben nach unten ausführen.
4. In den Grafik-Notebooks werden mit `save_figure = True` PNG- und PDF-Dateien gespeichert. Der Unterordner `fig/` wird bei Bedarf automatisch angelegt.

Ein fest eingestellter `seed` macht Simulationen reproduzierbar. Alle Notebooks trennen Eingaben, Prüfung, Berechnung, Grafik und Ausgabe sichtbar voneinander.
